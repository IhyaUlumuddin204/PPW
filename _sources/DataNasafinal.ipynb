#Sesi

```python
import pandas as pd

# Nama file input
file_name = "session_host_time_url_NASA.csv"

print(f"Memuat data dari {file_name}...")
df = pd.read_csv(file_name, sep=',')

# 1. Hanya ambil baris dengan url berakhir .html
target_column = "url"

if target_column not in df.columns:
    print(f"Error: Kolom '{target_column}' tidak ditemukan di CSV.")
else:
    print(f"\nMemfilter data untuk baris di mana '{target_column}' diakhiri dengan '.html'...")
    filtered_df = df[df[target_column].str.endswith('.html', na=False)].copy()

    # 2. Ubah kolom waktu ke datetime
    #    Sesuaikan nama kolom kalau beda (di screenshot: 'datetime_readable')
    filtered_df["datetime_readable"] = pd.to_datetime(filtered_df["datetime_readable"])

    # 3. Urutkan dulu berdasarkan host dan waktu
    filtered_df = filtered_df.sort_values(["host", "datetime_readable"])

    # 4. Buat kolom SESSION baru berdasarkan jeda > 1 menit per host
    session_threshold = pd.Timedelta(minutes=1)

    # diff() = selisih waktu dengan baris sebelumnya (untuk host yang sama)
    filtered_df["Sesi Login"] = (
        filtered_df
        .groupby("host")["datetime_readable"]
        .diff()                          # selisih waktu dengan baris sebelumnya
        .gt(session_threshold)           # True kalau selisih > 1 menit
        .fillna(True)                    # baris pertama tiap host dianggap sesi baru
        .groupby(filtered_df["host"])    # hitung kumulatif per host
        .cumsum()                        # 1,2,3,... per host
        .astype(int)
    )

    # Kalau kamu mau pakai nama kolom 'SESSION' saja (overwrite yang lama), bisa:
    # filtered_df.rename(columns={"SESSION_BARU": "SESSION"}, inplace=True)

    # 5. Simpan hasilnya
    output_filename = "final_filter_data_NASA_with_session.csv"
    filtered_df.to_csv(output_filename, sep=',', index=False)

    from IPython.display import display

    print(f"Memuat data dari {output_filename} untuk ditampilkan...")
    df_to_display = pd.read_csv(output_filename, sep=',')

    print("\nMenampilkan beberapa baris pertama dengan display():")
    display(df_to_display.head(30))
```

# Buat Abjad berdasarkan html unik

```python
from string import ascii_uppercase

# daftar huruf A-Z
letters = list(ascii_uppercase)

def assign_letter_per_host(group):
    # Ambil URL unik per host, urut sesuai kemunculan pertama
    unique_urls = group["url"].drop_duplicates()

    if len(unique_urls) > len(letters):
        print(f"Peringatan: host {group['host'].iloc[0]} punya lebih dari 26 halaman unik. "
              f"Hanya 26 pertama yang akan diberi huruf A-Z.")

    # Buat mapping: url -> huruf (A, B, C, ...)
    url_to_letter = {
        url: letters[i]
        for i, url in enumerate(unique_urls)
        if i < len(letters)              # jaga-jaga kalau >26
    }

    # Tambahkan kolom baru berisi huruf
    group["url_code"] = group["url"].map(url_to_letter)

    return group

# Terapkan per host
filtered_df = (
    filtered_df
    .sort_values(["host", "datetime_readable"])     # pastikan terurut dulu
    .groupby("host", group_keys=False)
    .apply(assign_letter_per_host)
)

# Kalau mau, kamu bisa drop kolom 'url' lama dan ganti jadi huruf,
# tapi biasanya lebih aman disimpan dua-duanya.
# Misalnya:
# filtered_df["url_asli"] = filtered_df["url"]
# filtered_df["url"] = filtered_df["url_code"]

# Simpan ke file baru
output_with_letters = "final_filter_data_NASA_with_session_and_letters.csv"
filtered_df.to_csv(output_with_letters, sep=",", index=False)

from IPython.display import display
print(f"\nMenampilkan beberapa baris pertama dengan kode huruf per host:")
display(filtered_df.head(30))
```

# Tabel Kunjungan Tiap host

```python
import pandas as pd

# pakai dataframe yang sudah ada url_code & sesi
df = filtered_df.copy()

# --- 1. Nama kolom sesi (ganti kalau beda) ---
session_col = "Sesi Login"   # kalau punyamu 'SESSION', ganti ke "SESSION"

# --- 2. Tambah kolom 'visited' berisi 1 ---
df["visited"] = 1

# --- 3. Pivot: 1 baris = host + sesi ---
wide_df = (
    df
    .pivot_table(
        index=["host", session_col],   # baris: per host per sesi
        columns="url_code",            # kolom: A, B, C, ...
        values="visited",
        aggfunc="max",                 # kalau ada beberapa baris, tetap 1
        fill_value=0                   # yang tidak dikunjungi → 0
    )
    .reset_index()
)

# hilangkan nama level kolom bawaan pivot_table
wide_df.columns.name = None

# --- 4. Simpan ke CSV baru ---
output_wide = "tabel_sesi_per_host.csv"
wide_df.to_csv(output_wide, sep=",", index=False)

from IPython.display import display
print("Beberapa baris pertama tabel sesi-per-host:")
display(wide_df.head(10))
```

# Tabel Kesimpulan

```python
import pandas as pd

# wide_df = tabel host + Sesi Login + kolom2 A..Z
df = wide_df.copy()

# 1. Tentukan kolom fitur (A,B,C,...)
feature_cols = [c for c in df.columns if c not in ["host", "Sesi Login"]]

# 2. Hitung berapa banyak halaman (A..Z) yang dikunjungi di tiap sesi
df["total_visited"] = df[feature_cols].sum(axis=1)

# 3. Untuk tiap host, pilih sesi yang:
#    - total_visited paling besar (kunjungan terbanyak)
#    - kalau seri (imbang), ambil Sesi Login terkecil (paling awal)
best_sessions = (
    df
    .sort_values(["host", "total_visited", "Sesi Login"], ascending=[True, False, True])
    .drop_duplicates(subset="host")   # sisain 1 sesi terbaik per host
)

# 4. Ambil hanya host + pola A..Z
summary_df = best_sessions[["host"] + feature_cols].reset_index(drop=True)

# 5. Simpan ke CSV
output_summary = "tabel_host_sesi_dominan.csv"
summary_df.to_csv(output_summary, sep=",", index=False)

from IPython.display import display
print("Tabel host dengan sesi paling dominan (jumlah kunjungan terbanyak):")
display(summary_df.head(10))
```

```python
import pandas as pd

# pakai dataframe yang sudah ada kolom 'host' dan 'url_code'
df = filtered_df.copy()

# 1. Anggap setiap baris = 1 kunjungan
df["visited"] = 1

# 2. Hilangkan duplikat host+url_code (biar tiap host-url_code cuma sekali)
df_unique = df.drop_duplicates(subset=["host", "url_code"])

# 3. Pivot: 1 baris = 1 host, kolom = A,B,C,... berisi 1/0
host_summary = (
    df_unique
    .pivot_table(
        index="host",          # baris: per host
        columns="url_code",    # kolom: A, B, C, ...
        values="visited",
        aggfunc="max",         # kalau ada, jadinya 1
        fill_value=0           # kalau tidak ada, 0
    )
    .reset_index()
)

# Bersihkan nama level kolom bawaan pivot_table
host_summary.columns.name = None

# 4. Simpan ke CSV
output_host = "tabel_total_kunjungan_per_host.csv"
host_summary.to_csv(output_host, sep=",", index=False)

from IPython.display import display
print("Beberapa baris pertama total kunjungan per host:")
display(host_summary.head(10))
```
