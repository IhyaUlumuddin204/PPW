```python
import pandas as pd
df = pd.read_excel("webuage.xlsx")
display(df.head())
```

```python
html_requests = df[df['Request URI'].str.contains('.html', na=False)]
count_html_requests = len(html_requests)
print(f"Jumlah Request URI yang mengandung '.html': {count_html_requests}")
```

```python
# Mulai dengan memfilter permintaan yang mengandung '.html' di 'Request URI'
filtered_requests = df[df['Request URI'].str.contains('.html', na=False)].copy()

# Kemudian, filter lebih lanjut untuk 'GET' dalam kolom 'Request method'
# Menggunakan kolom 'Request method' untuk memfilter permintaan 'GET'.
filtered_requests = filtered_requests[
    filtered_requests['Request method'] == 'GET'
].copy()
```

```python
# Buat kolom 'time' dari 'Request time' dengan format lengkap (tahun, bulan, hari, jam, menit, detik)
filtered_requests['time'] = pd.to_datetime(filtered_requests['Request time'], format='ISO8601').dt.strftime('%Y-%m-%d %H:%M:%S')

# Buat kolom 'ip' dari 'Remote host'
filtered_requests['ip'] = filtered_requests['Remote host']

# Buat kolom 'URL' dari 'Request URI'
filtered_requests['URL'] = filtered_requests['Request URI']
```

```python
# Filter untuk Status 200 saja
filtered_requests_status_200 = filtered_requests[filtered_requests['Status'] == 200].copy()

# Pilih kolom yang diinginkan dan tambahkan kolom 'status' dan 'Size'
final_output_df = filtered_requests_status_200[['time', 'ip', 'URL']].copy()
final_output_df['status'] = filtered_requests_status_200['Status']
final_output_df['Size'] = filtered_requests_status_200['Size of response (incl. headers)']
final_output_df['Method'] = filtered_requests_status_200['Request method']

# Tampilkan 5 baris pertama dari hasil
display(final_output_df.head())

# Simpan hasilnya ke file Excel
output_filename = 'filtered_web_requests.xlsx'
final_output_df.to_excel(output_filename, index=False)

print(f"Data yang difilter telah disimpan ke '{output_filename}'")
```

```python
final_output_df = final_output_df.sort_values(by=['ip', 'time'], ascending=[True, True])
display(final_output_df.head())
```

```python
output_filename_sorted = 'hasil_filter.xlsx'
final_output_df.to_excel(output_filename_sorted, index=False)
print(f"Data yang diurutkan telah disimpan ke '{output_filename_sorted}'")
```
